# Assignment 03 - Pie Chart

```python
# Importing required libraries 
import pandas as pd 
import matplotlib.pyplot as plt
```

## Loading the CSV

```python
url = "https://drive.google.com/uc?id=1AWPf-pJodJKeHsARQK_RHiNsE8fjPCVK&export=download" 

df = pd.read_csv(url)
```

## Inspecting the Data and Structure

```python
df.head ()
```

```python
df.info()
```

## Splitting the Email Domains

```python
df["domain"] = df["Email"].str.split("@").str[1]
```

## Count of Domains

```python
domain_counts = df["domain"].value_counts()
```

```python
print(domain_counts)
```

## Creating the Pie Chart

```python
plt.pie(domain_counts) 

plt.show()
```

## Adding Labels and Percentages

```python
plt.pie(
    domain_counts, 
    labels=domain_counts.index
)
```

```python
plt.pie(
    domain_counts,
    labels=domain_counts.index,
    autopct="%1.1f%%"
)
```

## Rotation, Colours and Title

```python
plt.pie(
    domain_counts,
    labels=domain_counts.index,
    autopct="%1.1f%%",
    startangle=90
)
```

```python
colors = plt.cm.Set3.colors
```

```python
plt.pie(
    domain_counts,
    labels=domain_counts.index,
    autopct="%1.1f%%",
    startangle=90,
    colors=colors
)
```

```python
plt.title("Distribution of Email Domains")
```

## Final Pie Chart

```python
plt.figure(figsize=(6, 6))

plt.pie(
    domain_counts,
    labels=domain_counts.index,
    autopct="%1.1f%%",
    startangle=90,
    colors=plt.cm.Set3.colors
)

plt.title("Distribution of Email Domains")

plt.show()
```
