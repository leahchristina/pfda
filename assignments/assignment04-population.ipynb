## Part 1 - The Differences Between the Sexes by Age

```python
ireland_sex_age = population[
    (population["Administrative Counties"] == "Ireland") &
    (population["Sex"] != "Both sexes") &
    (population["Single Year of Age"] != "All ages")
].copy()
```

```python
ireland_sex_age.head()
```

```python
# Ensuring the total values have been removed as to not skew data 
print(ireland_sex_age["Administrative Counties"].unique())
print(ireland_sex_age["Sex"].unique())
print(ireland_sex_age["Single Year of Age"].unique())
```

```python
# Changing the age data into numerical
ireland_sex_age["Age"] = ireland_sex_age["Single Year of Age"]

ireland_sex_age["Age"] = ireland_sex_age["Age"].replace({
    "Under 1 year": "0",
    "100 years and over": "100"
})

ireland_sex_age["Age"] = (
    ireland_sex_age["Age"]
    .str.replace(" years", "", regex=False)
    .str.replace(" year", "", regex=False)
)

ireland_sex_age["Age"] = ireland_sex_age["Age"].astype(int)
```

```python
# Checking result 
ireland_sex_age[
    ["Single Year of Age", "Age", "Sex"]
].head(10)
```

```python
# Checking if VALUE column is numeric 
ireland_sex_age["VALUE"].dtype
```

```python
# Creating the weighted value
ireland_sex_age["Age_x_Population"] = (
    ireland_sex_age["Age"] * ireland_sex_age["VALUE"]
)
```

```python
ireland_sex_age[
    ["Sex", "Age", "VALUE", "Age_x_Population"]
].head(10)
```

```python
# Weighted Mean for Each Sex
weighted_mean_age = (
    ireland_sex_age
    .groupby("Sex")
    .apply(
        lambda group:
        group["Age_x_Population"].sum()
        / group["VALUE"].sum()
    )
)

weighted_mean_age
```

```python
# Make the data easier to read

ireland_sex_age["Age_x_Population"] = (
    ireland_sex_age["Age"] * ireland_sex_age["VALUE"]
)

weighted_mean_age = (
    ireland_sex_age
    .groupby("Sex")
    .apply(
        lambda group:
        group["Age_x_Population"].sum()
        / group["VALUE"].sum()
    )
    .reset_index(name="Weighted Mean Age")
)

weighted_mean_age["Weighted Mean Age"] = (
    weighted_mean_age["Weighted Mean Age"].round(2)
)

weighted_mean_age
```

## Conclusion 
The weighted mean age was calculated by multiplying each age by the popu;ation of that age, summing those values for each sex and dividing by the total population.
For females the weighted mean age is 39 (rounded) and males is 38 (rounded), showing little population distribution changes between the two sexes.
One limitation of the dataset is that for individuals over 100 years, they are not represented by their exact age. This could slightly unstate the weighted mean. 

# Part 2: Creating a 5-Year Group

```python
# Store the age that will be examined
selected_age = 25
```

```python
# Include people within five years either side of the selected age 
lower_age = selected_age - 5
upper_age = selected_age + 5
```

```python
# Filter the Ireland data to the selected age range
selected_age_group = ireland_sex_age[
    ireland_sex_age["Age"].between(lower_age, upper_age)
].copy()

# Add the population values for each sex
age_group_totals = (
    selected_age_group
    .groupby("Sex")["VALUE"]
    .sum()
)

# Calculate the difference between the sexes
population_difference = (
    age_group_totals["Female"] - age_group_totals["Male"]
)

print(f"Selected age: {selected_age}")
print(f"Age range: {lower_age} to {upper_age}")
print()
print(age_group_totals)
print()

if population_difference > 0:
    print(
        f"There are {population_difference:,} more females than males "
        f"aged {lower_age} to {upper_age} in Ireland."
    )
elif population_difference < 0:
    print(
        f"There are {abs(population_difference):,} more males than females "
        f"aged {lower_age} to {upper_age} in Ireland."
    )
else:
    print(
        f"The male and female populations aged {lower_age} to "
        f"{upper_age} in Ireland are equal."
    )
```

## Conclusion
The selected age is stored in the variable selected_age and can be updated, for this example I have chosen 25. 
People within 5 years of either side of 25 are grouped into an age range, therefore the age range is 20 - 30 inclusive.
The population difference is calculated as = Female population - male population. A positive result means there are more females, while a negative result means there are more males. From the example above, we can see there are 754 more males than females in Ireland.

The data in the dataset is displayed under County Councils for each area. To sort into regions, I'm using NUTS3["https://en.wikipedia.org/wiki/NUTS_statistical_regions_of_Ireland"] structure, which sorts the data into eight statistical regions.

```python
region_mapping = {
    # Border
    "Cavan County Council": "Border",
    "Donegal County Council": "Border",
    "Leitrim County Council": "Border",
    "Monaghan County Council": "Border",
    "Sligo County Council": "Border",

    # West
    "Galway City Council": "West",
    "Galway County Council": "West",
    "Mayo County Council": "West",
    "Roscommon County Council": "West",

    # Mid-West
    "Clare County Council": "Mid-West",
    "Limerick City & County Council": "Mid-West",
    "Tipperary County Council": "Mid-West",

    # South-East
    "Carlow County Council": "South-East",
    "Kilkenny County Council": "South-East",
    "Waterford City & County Council": "South-East",
    "Wexford County Council": "South-East",

    # South-West
    "Cork City Council": "South-West",
    "Cork County Council": "South-West",
    "Kerry County Council": "South-West",

    # Dublin
    "Dublin City Council": "Dublin",
    "Dún Laoghaire Rathdown County Council": "Dublin",
    "Fingal County Council": "Dublin",
    "South Dublin County Council": "Dublin",

    # Mid-East
    "Kildare County Council": "Mid-East",
    "Louth County Council": "Mid-East",
    "Meath County Council": "Mid-East",
    "Wicklow County Council": "Mid-East",

    # Midland
    "Laois County Council": "Midland",
    "Longford County Council": "Midland",
    "Offaly County Council": "Midland",
    "Westmeath County Council": "Midland"
}
```

```python
# Creating a copy for a separate regional DataFrame

regional_population = population.copy()

regional_population["Region"] = (
    regional_population["Administrative Counties"]
    .map(region_mapping)
)
```

```python
# Checking my mapping - this is updated to remove spacing as 'Mayo County Council' has two spaces between the Mayo and County, causing an NaN in my previous try.

regional_population["Administrative Counties"] = (
    regional_population["Administrative Counties"]
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)

regional_population["Region"] = (
    regional_population["Administrative Counties"]
    .map(region_mapping)
)
```
